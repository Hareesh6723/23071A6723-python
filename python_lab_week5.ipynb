```python
# ii) Write a python program to perform addition of two square matrices
r = int(input("Enter number of rows: "))
c = int(input("Enter number of columns: "))
m1 = []
m2 = []

print("Enter elements for the first matrix:")
for i in range(r):
    a = []
    for j in range(c):
        a.append(int(input()))
    m1.append(a)

print("Enter elements for the second matrix:")
for i in range(r):
    b = []
    for j in range(c):
        b.append(int(input()))
    m2.append(b)

m3 = []
for i in range(r):
    d = []
    for j in range(c):
        d.append(m1[i][j] + m2[i][j])
    m3.append(d)
print(m3)
```

```python
# iii) Write a python program to perform multiplication of two square matrices
r = int(input("Enter number of rows: "))
c = int(input("Enter number of columns: "))
m1 = []
m2 = []

print("Enter elements for the first matrix:")
for i in range(r):
    a = []
    for j in range(c):
        a.append(int(input()))
    m1.append(a)

print("Enter elements for the second matrix:")
for i in range(r):
    b = []
    for j in range(c):
        b.append(int(input()))
    m2.append(b)

m3 = []
for i in range(r):
    d = []
    for j in range(c):
        s=0
        for k in range(r):
                s=s+(m1[i][k] * m2[k][j])
        d.append(s)
    m3.append(d)
print(m3)
```

```python
# 2 Simple Calculator program by making use of functions
def add(a,b):
    return a+b
def sub(a,b):
    return a-b
def mult(a,b):
    return a*b
def div(a,b):
    return a/b
a=input()
if a=="+" :
     r= add(int(input()),int(input()))
elif a =="-" :
     r= sub(int(input()),int(input()))
elif a=="*" :
     r= mult(int(input()),int(input()))
else :
     r=div(int(input()),int(input()))
print(r)

        
```

```python
#3. Find the factorial of a number using recursion
def fact(n):
    if(n==0):
        return 1
    else:
        return fact(n-1)*n
r=fact(int(input()))
print(r)
```

```python
#4. Write a function cumulative_ product to compute cumulative product of a list of numbers.

l=list(map(int,input().split(" ")))
r=[]
s=1
for i in range(len(l)):
    s=s*l[i]
    r.append(s)
print(r)
        
```

```python
# 5. Write a function reverse to print the given list in the reverse order. 
def rever(l):
    l1=[]
    for i in range(len(l)-1,-1,-1):
        l1.append(l[i])
    return l1
r=rever(list(map(int,input().split(" "))))
print(r)
```
